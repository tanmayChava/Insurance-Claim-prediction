### Convert DataType

#### By default, pandas loads numbers as int64 or float64, which are heavy.

#### Example:
#### int64 → 8 bytes
#### int8 → 1 byte

#### If a column only has values 0/1 (binary), storing it as int64 wastes 7 extra bytes per row.
####  Converting reduces RAM usage & speeds up processing.

```python
print(data["ps_ind_06_bin"].unique())
```

#### 1) Binary (*_bin)
#### Values: 0/1 → can safely use int8.

#### 2) Categorical (*_cat)
#### These are labels, not numbers → use category (pandas) or encode later for ML.

#### 3) Numeric (continuous)
#### Columns like ps_reg_01, ps_car_13, ps_calc_01 etc.
#### Keep as float32 (lighter than float64, but still precise enough).

#### 4) ID / Target
#### id → usually drop (not predictive).
#### target → keep as int8 (0/1).

```python
cat_cols = [c for c in data.columns if "_cat" in c]
bin_cols = [c for c in data.columns if "_bin" in c]
num_cols = [c for c in data.columns if c not in (cat_cols + bin_cols + ["id","target"])]

data[cat_cols] = data[cat_cols].astype("category")
data[bin_cols] = data[bin_cols].astype("int8")
data[num_cols] = data[num_cols].astype("float32")
data["target"] = data["target"].astype("int8")
```

## Handling Outliers

### View Outliers

```python
import pandas as pd

# Simple outlier check using IQR
num_cols = data.select_dtypes(include=["int64","float64","float32"]).columns
outlier_counts = {}

for col in num_cols:
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = ((data[col] < lower) | (data[col] > upper)).sum()
    if outliers > 0:
        outlier_counts[col] = outliers

# Show columns with outliers, sorted
outlier_counts = pd.Series(outlier_counts).sort_values(ascending=False)
print(outlier_counts)
```

### Handle Outliers

```python
for col in data.select_dtypes(include=["int64","float64"]).columns:
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    data[col] = data[col].clip(lower, upper)
```

### Repalcing -1 with NAN

#### We replace -1 with NaN because -1 is just a placeholder for “missing”, while NaN is the proper way to represent missing values in ML workflows. This prevents the model from misinterpreting -1 as a real numeric value.

```python
import numpy as np

# Replace -1 with NaN
data = data.replace(-1, np.nan)

# Fill NaN with median (better than mean for skewed data)
data = data.fillna(data.median(numeric_only=True))
```

```python
# Numeric imputation
num_cols = data.select_dtypes(include=["int64","float64"]).columns
data[num_cols] = data[num_cols].apply(lambda x: x.fillna(x.median()))

# Categorical imputation
cat_cols = [col for col in data.columns if "_cat" in col or "_bin" in col]
for col in cat_cols:
    data[col] = data[col].fillna(data[col].mode()[0])
```

## Model Creation

### Defining X & Y

```python
X = data.drop(columns = ['target'])
y = data['target']
```

### Train - Test Split

```python
from sklearn.model_selection import train_test_split
X_train , X_test , y_train , y_test = train_test_split(X,y, test_size = 0.2, stratify=y, random_state=42)
print(X_train.shape, X_test.shape)
```

### Using Smote

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)
X_train_res, y_train_res = smote.fit_resample(X_train, y_train)

print("Before SMOTE:", y_train.value_counts())
print("After SMOTE:", y_train_res.value_counts())
```

## Model Training

### Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score

# 1. Initialize Random Forest
rf = RandomForestClassifier(
    n_estimators=200,      # number of trees
    max_depth=None,        # let trees grow fully (can tune later)
    random_state=42,
    n_jobs=-1,             # use all CPU cores
    class_weight="balanced"      # not needed since we used SMOTE
)

# 2. Train on SMOTE-resampled training data
rf.fit(X_train_res, y_train_res)

# 3. Predictions on test set
y_pred = rf.predict(X_test)
y_proba = rf.predict_proba(X_test)[:, 1]  # probabilities for ROC-AUC

# 4. Evaluation
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
print("\nROC-AUC Score:", roc_auc_score(y_test, y_proba))
```

#### Class 0 (majority) → almost perfect (recall = 1.00, precision = 0.96).

#### Class 1 (minority, the one we care about) → completely ignored (precision = 0.00, recall = 0.00, F1 = 0.00).

#### ROC-AUC ≈ 0.59 → barely better than random guessing (0.5).

#### 👉 Even though overall accuracy looks high (96%), the model is useless for detecting class 1.

```python
# Convert categorical columns to int
cat_cols = [col for col in X_train.columns if "_cat" in col]
X_train[cat_cols] = X_train[cat_cols].astype("int")
X_test[cat_cols] = X_test[cat_cols].astype("int")
```

### Xgboost

```python
from xgboost import XGBClassifier
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score

# Compute class weight ratio
scale_pos_weight = (y_train == 0).sum() / (y_train == 1).sum()
print("Scale_pos_weight:", scale_pos_weight)

# Initialize XGBoost
xgb = XGBClassifier(
    n_estimators=500,
    learning_rate=0.05,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=scale_pos_weight,  # <-- handle imbalance
    random_state=42,
    n_jobs=-1,
    eval_metric="auc"
)

# Train
xgb.fit(X_train, y_train)

# Predictions
y_pred = xgb.predict(X_test)
y_proba = xgb.predict_proba(X_test)[:, 1]

# Evaluation
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
print("\nROC-AUC Score:", roc_auc_score(y_test, y_proba))
```

#### 📈 Classification Report

#### Class 0 (majority)

#### Precision = 0.97 → when it predicts 0, it’s correct most of the time.

#### Recall = 0.73 → it misses ~27% of actual 0s.

#### Class 1 (minority, the important one)

#### Precision = 0.06 → only 6% of predicted 1s are actually correct → many false positives.

#### Recall = 0.44 → model finds 44% of real positives, which is much better than before (Random Forest had 0.00 recall).

#### Macro avg F1 = 0.47 → overall balance between the two classes is still low.

## Conclusion

#### The results demonstrate that XGBoost outperforms Random Forest when dealing with imbalanced data, particularly in capturing the minority class. While Random Forest completely failed to identify class 1, XGBoost achieved a recall of 44% for the minority class, albeit with low precision. This trade-off highlights the importance of threshold tuning in imbalanced classification problems.

#### However, due to the anonymized nature of the dataset, we cannot interpret which features are most influential or provide business-level insights. Future improvements would require feature documentation to connect model outputs with real-world meaning.
